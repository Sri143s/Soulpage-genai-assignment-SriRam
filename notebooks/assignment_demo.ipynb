## 💬 Task 2: Conversational Knowledge Bot

```python
from task2_knowledge_bot.bot import knowledge_bot
from task2_knowledge_bot.memory import memory_manager

session_id = "notebook_chat_session"
memory_manager.clear_history(session_id)

# Turn 1: Initial Question
q1 = "Who is the CEO of OpenAI?"
res1 = knowledge_bot.ask(q1, session_id=session_id)
print(f"Q: {q1}")
print(f"A: {res1['response']}\n")

# Turn 2: Follow-up Question with Pronoun 'he'
q2 = "Where did he study?"
res2 = knowledge_bot.ask(q2, session_id=session_id)
print(f"Q: {q2}")
print(f"Resolved Query: '{res2.get('resolved_query')}'")
print(f"A: {res2['response']}")
```
